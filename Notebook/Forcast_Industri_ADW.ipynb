```python
print("Ukuran Dataset:", df.shape)

print("\nInformasi Dataset:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nJumlah Duplikasi:")
print(df.duplicated().sum())

print("\nLima Data Pertama:")
display(df.head())
```
